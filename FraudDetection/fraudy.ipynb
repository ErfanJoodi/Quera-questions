```python
import numpy as np
import pandas as pd
```

```python
train= pd.read_csv(r"F:\Programming practices\Quera questions\FraudDetection\Data\train.csv")
test= pd.read_csv(r"F:\Programming practices\Quera questions\FraudDetection\Data\test.csv")
```

    تکنیک‌هایی که استفاده می‌کنید به شکل مستقیم مورد ارزیابی توسط سامانه داوری قرار <b>نمی‌گیرند.</b> بلکه همه آن‌ها در دقت مدل شما تاثیر خواهند گذاشت؛ بنابراین هر چه پیش‌پردازش/مهندسی ویژگی بهتری انجام دهید تا دقت مدل بهبود پیدا کند، امتیاز بیشتری از این سوال کسب خواهید کرد.
    در این قسمت شما می‌توانید بخشی از داده‌ی موجود را برای اعتبارسنجی در نظر بگیرید.

```python
from sklearn.preprocessing import OrdinalEncoder, StandardScaler
from sklearn.model_selection import train_test_split
```

```python
df= train.copy()
X_test= test.copy()

df.drop(columns= ['timestamp', 'user_id', 'processing_batch_id'], inplace= True)
X_test.drop(columns= ['timestamp', 'user_id', 'processing_batch_id'], inplace= True)

X= df.drop(columns= ['is_fraud']).copy()
y= df['is_fraud'].copy()

encoder= OrdinalEncoder(handle_unknown= 'use_encoded_value', unknown_value= -1)
cat_cols= ['product_category', 'payment_method']
X[cat_cols]= encoder.fit_transform(X[cat_cols])
X_test[cat_cols]= encoder.transform(X_test[cat_cols])

scaler= StandardScaler()
s_cols= X.columns[: -2].to_list()
X[s_cols]= scaler.fit_transform(X[s_cols])
X_test[s_cols]= scaler.transform(X_test[s_cols])

X_train, X_val, y_train, y_val= train_test_split(X, y, test_size= 0.14, random_state= 14, stratify= y)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
lr_model= LogisticRegression(max_iter= 1000, random_state= 48)
lr_model.fit(X_train, y_train)
predict= lr_model.predict(X_val)
accuracy_score(y_val, predict)
```

```python
import zipfile
import joblib
import os

if not os.path.exists(os.path.join(os.getcwd(), 'fraudy.ipynb')):
    %notebook -e fraudy.ipynb

def compress(file_names):
    print("File Paths:")
    print(file_names)
    compression = zipfile.ZIP_DEFLATED
    with zipfile.ZipFile("result.zip", mode="w") as zf:
        for file_name in file_names:
            zf.write('./' + file_name, file_name, compress_type=compression)

submission.to_csv('submission.csv', index=False)
file_names = ['fraudy.ipynb', 'submission.csv']
compress(file_names)
```
