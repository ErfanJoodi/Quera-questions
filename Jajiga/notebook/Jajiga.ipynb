```python
train= pd.read_csv(r"F:\Programming practices\Quera questions\Jajiga\data\train.csv")
test= pd.read_csv(r"F:\Programming practices\Quera questions\Jajiga\data\test.csv")
```

    <br>
    تکنیک‌هایی که استفاده می‌کنید به شکل مستقیم مورد ارزیابی توسط سامانه داوری قرار <b>نمی‌گیرند.</b> بلکه همه آن‌ها در دقت مدل شما تاثیر خواهند گذاشت.

```python
from sklearn.preprocessing import MultiLabelBinarizer, OrdinalEncoder, StandardScaler
from sklearn.model_selection import train_test_split
```

```python
df= train.copy()
X_test= test.copy()

df['sleep_arrange_len']= df['sleep_arrange'].apply(lambda x: len(eval(x)))
df['sleep_arrange_sum_single']= df['sleep_arrange'].apply(lambda x: np.sum([i['single'] for i in eval(x)]))
df['sleep_arrange_sum_double']= df['sleep_arrange'].apply(lambda x: np.sum([i['double'] for i in eval(x)]))
df['sleep_arrange_sum_mattress']= df['sleep_arrange'].apply(lambda x: np.sum([i['mattress'] for i in eval(x)]))
X_test['sleep_arrange_len']= X_test['sleep_arrange'].apply(lambda x: len(eval(x)))
X_test['sleep_arrange_sum_single']= X_test['sleep_arrange'].apply(lambda x: np.sum([i['single'] for i in eval(x)]))
X_test['sleep_arrange_sum_double']= X_test['sleep_arrange'].apply(lambda x: np.sum([i['double'] for i in eval(x)]))
X_test['sleep_arrange_sum_mattress']= X_test['sleep_arrange'].apply(lambda x: np.sum([i['mattress'] for i in eval(x)]))

df.drop(columns= ['Unnamed: 0', 'id', 'status', 'title', 'description', 'floors_count', 'sleep_arrange', 'sleep_description', 'guest_number', 'stays_max', 'additional_feature', 'additional_safety', 'additional_rule', 'province.name', 'province.url', 'city.name', 'city.url', 'units_count'], inplace= True)
X_test.drop(columns= ['Unnamed: 0', 'id', 'status', 'title', 'description', 'floors_count', 'sleep_arrange', 'sleep_description', 'guest_number', 'stays_max', 'additional_feature', 'additional_safety', 'additional_rule', 'province.name', 'province.url', 'city.name', 'city.url', 'units_count'], inplace= True)

df['features']= df['features'].apply(lambda x: [i['name'] for i in eval(x)])
X_test['features']= X_test['features'].apply(lambda x: [i['name'] for i in eval(x)])

for col in ['types', 'regions', 'rules', 'features']:
    df[col]= df[col].apply(lambda x: list(map(str, x)))
    mlb= MultiLabelBinarizer()
    mocodes_encoded= mlb.fit_transform(df[col])
    mocodes_df= pd.DataFrame(mocodes_encoded, columns= [f'Mocode_{c}' for c in mlb.classes_], index= df.index)
    df= pd.concat([df.drop(columns= [col]), mocodes_df], axis= 1)
    X_test[col]= X_test[col].apply(lambda x: list(map(str, x)))
    # mlb= MultiLabelBinarizer()
    mocodes_encoded= mlb.transform(X_test[col])
    mocodes_df= pd.DataFrame(mocodes_encoded, columns= [f'Mocode_{c}' for c in mlb.classes_], index= X_test.index)
    X_test= pd.concat([X_test.drop(columns= [col]), mocodes_df], axis= 1)

for col in df.select_dtypes(include= ['bool']).columns:
    df[col]= df[col].apply(lambda x: 1 if x==True else 0)
```

```python
X= df.drop(columns= ['min_price'])
y= df['min_price']

X_train, X_val, y_train, y_val= train_test_split(X, y, random_state= 48, test_size= 0.2)

cat_cols= X.select_dtypes(include= ['object', 'category']).columns
le= OrdinalEncoder()
X_train[cat_cols]= le.fit_transform(X_train[cat_cols])
X_val[cat_cols]= le.transform(X_val[cat_cols])
scaler= StandardScaler()
X_train= pd.DataFrame(scaler.fit_transform(X_train), columns= X_train.columns, index= X_train.index)
X_val= pd.DataFrame(scaler.transform(X_val), columns= X_val.columns, index= X_val.index)
```

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor, VotingRegressor
from xgboost import XGBRegressor
from sklearn.metrics import r2_score
import warnings
```

```python
warnings.filterwarnings('ignore')
```

```python
lr_model= LinearRegression()
lr_model.fit(X_train, y_train)
predict= lr_model.predict(X_val)
r2_score(y_val, predict)
```

```python
rid_model= Ridge(random_state= 48)
param_grid= {
    'alpha': [0.1, 1, 10, 100],
    'max_iter': [10, 100, 1000]
}
grid= GridSearchCV(rid_model, param_grid, cv= 4, scoring= 'r2', n_jobs= -1, verbose= 2)
grid.fit(X_train, y_train)
print(grid.best_params_)
predict= grid.best_estimator_.predict(X_val)
r2_score(y_val, predict)
```

```python
las_model= Lasso(random_state= 48)
param_grid= {
    'alpha': [0.1, 1, 10, 100],
    'max_iter': [10, 100, 1000]
}
grid= GridSearchCV(las_model, param_grid, cv= 4, scoring= 'r2', n_jobs= -1, verbose= 2)
grid.fit(X_train, y_train)
print(grid.best_params_)
predict= grid.best_estimator_.predict(X_val)
r2_score(y_val, predict)
```

```python
knn_model= KNeighborsRegressor()
param_grid= {
    'n_neighbors': [3, 5, 7]
}
grid= GridSearchCV(knn_model, param_grid, scoring= 'r2', cv= 4, n_jobs= -1, verbose= 2)
grid.fit(X_train, y_train)
print(grid.best_params_)
predict= grid.best_estimator_.predict(X_val)
r2_score(y_val, predict)
```

```python
svr_model= SVR()
param_grid= {
    'kernel': ['linear', 'rbf', 'sigmoid'],
    'gamma': [0.1, 1, 10],
    'C': [0.01, 0.1, 1]
}
grid= GridSearchCV(svr_model, param_grid, scoring= 'r2', cv= 4, n_jobs= -1, verbose= 2)
grid.fit(X_train, y_train)
print(grid.best_params_)
predict= grid.best_estimator_.predict(X_val)
r2_score(y_val, predict)
```

```python
rf_model= RandomForestRegressor(random_state= 48)
param_grid= {
    'n_estimators': [10, 100, 1000],
    'max_depth': [5, 10, 100]
    }
grid= GridSearchCV(rf_model, param_grid, cv= 4, scoring= 'r2', n_jobs= -1, verbose= 2)
grid.fit(X_train, y_train)
print(grid.best_params_)
predict= grid.best_estimator_.predict(X_val)
r2_score(y_val, predict)
```

```python
xgb_model= XGBRegressor(random_state= 48)
param_grid= {
    'n_estimators': [10, 100, 1000],
    'max_depth': [5, 10, 100],
    'learning_rate': [0.01, 0.1]
}
grid= GridSearchCV(xgb_model, param_grid, cv= 4, scoring= 'r2', n_jobs=-1, verbose= 2)
grid.fit(X_train.to_numpy(), y_train)
print(grid.best_params_)
predict= grid.best_estimator_.predict(X_val.to_numpy())
r2_score(y_val, predict)
```

```python
ensemble= VotingRegressor([
    ('ridge', LinearRegression()),
    # ('knn', KNeighborsRegressor(n_neighbors= 7)),
    # ('svm', SVR(kernel= 'linear', gamma= 0.1, C= 1)),
    ('rf', RandomForestRegressor(n_estimators= 1000, max_depth= 100, random_state= 48)),
    # ('xgb', XGBRegressor(learning_rate=0.1, max_depth=10, n_estimators=500, random_state=48))
])
ensemble.fit(X_train, y_train)
predict= ensemble.predict(X_val)
r2_score(y_val, predict)
```
