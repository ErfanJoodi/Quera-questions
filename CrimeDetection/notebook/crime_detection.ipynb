```python
import pandas as pd
import numpy as np
import warnings
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import MultiLabelBinarizer
from sklearn.feature_selection import SelectKBest, chi2
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score
from sklearn.ensemble import RandomForestClassifier
```

```python
warnings.filterwarnings('ignore')
```

```python
train= pd.read_csv(r"F:\Programming practices\Quera questions\CrimeDetection\data\train.csv")
test= pd.read_csv(r"F:\Programming practices\Quera questions\CrimeDetection\data\test.csv")
```

```python
df= train.copy()
X_test= test.copy()

df.drop(columns= ['AREA NAME', 'Rpt Dist No', 'LOCATION', 'Cross Street', 'LAT', 'LON', 'Crm Cd Desc', 'Crm Cd 1', 'Crm Cd 2', 'Crm Cd 3', 'Crm Cd 4'], inplace= True)
X_test.drop(columns= ['AREA NAME', 'Rpt Dist No', 'LOCATION', 'Cross Street', 'LAT', 'LON'], inplace= True)

df['Date Rptd']= pd.to_datetime(df['Date Rptd'])
X_test['Date Rptd']= pd.to_datetime(X_test['Date Rptd'])

df['DATE OCC']= pd.to_datetime(df['DATE OCC'])
X_test['DATE OCC']= pd.to_datetime(X_test['DATE OCC'])

df['Premis Cd'].fillna(0, inplace= True)
df['Premis Cd']= df['Premis Cd'].astype(int)
X_test['Premis Cd'].fillna(0, inplace= True)
X_test['Premis Cd']= X_test['Premis Cd'].astype(int)

df['Weapon Used Cd'].fillna(0, inplace= True)
df['Weapon Used Cd']= df['Weapon Used Cd'].astype(int)
X_test['Weapon Used Cd'].fillna(0, inplace= True)
X_test['Weapon Used Cd']= X_test['Weapon Used Cd'].astype(int)

df['Vict Sex'].fillna(df['Vict Sex'].mode()[0], inplace= True)
X_test['Vict Sex'].fillna(df['Vict Sex'].mode()[0], inplace= True)

df['Vict Descent'].fillna(df['Vict Descent'].mode()[0], inplace= True)
X_test['Vict Descent'].fillna(df['Vict Descent'].mode()[0], inplace= True)

for col in df.select_dtypes(include= ['datetime64']).columns:
    df[col + '_year']= df[col].dt.year
    df[col + '_month']= df[col].dt.month
    df[col + '_day']= df[col].dt.day
    df.drop(columns= [col], inplace= True)
    X_test[col + '_year'] = X_test[col].dt.year
    X_test[col + '_month'] = X_test[col].dt.month
    X_test[col + '_day'] = X_test[col].dt.day
    X_test.drop(columns= [col], inplace= True)

df['Mocodes']= df['Mocodes'].fillna('').apply(lambda x: x.split() if x != '' else [])
mlb= MultiLabelBinarizer()
mocodes_encoded= mlb.fit_transform(df['Mocodes'])
mocodes_df= pd.DataFrame(mocodes_encoded, columns= [f'Mocode_{c}' for c in mlb.classes_], index= df.index)
df= pd.concat([df.drop(columns= ['Mocodes']), mocodes_df], axis= 1)
X_test['Mocodes']= X_test['Mocodes'].fillna('').apply(lambda x: x.split() if x != '' else [])
# mlb= MultiLabelBinarizer()
mocodes_encoded= mlb.transform(X_test['Mocodes'])
mocodes_df= pd.DataFrame(mocodes_encoded, columns= [f'Mocode_{c}' for c in mlb.classes_], index= X_test.index)
X_test= pd.concat([X_test.drop(columns= ['Mocodes']), mocodes_df], axis= 1)

for col in ['Premis Desc', 'Weapon Desc', 'Status Desc']:
    tfidf= TfidfVectorizer(max_features= 4000, ngram_range= (1, 2), stop_words= 'english')
    df[col].fillna('No Desc', inplace= True)
    tfidf_matrix= tfidf.fit_transform(df[col])
    tfidf_df= pd.DataFrame(tfidf_matrix.toarray(), columns= tfidf.get_feature_names_out())
    df= pd.concat([df.drop(columns= [col]).reset_index(drop= True), tfidf_df.reset_index(drop= True)], axis= 1)
    # tfidf= TfidfVectorizer(max_features= 4000, ngram_range= (1, 2), stop_words= 'english')
    X_test[col].fillna('No Desc', inplace= True)
    tfidf_matrix= tfidf.transform(X_test[col])
    tfidf_df= pd.DataFrame(tfidf_matrix.toarray(), columns= tfidf.get_feature_names_out())
    X_test= pd.concat([X_test.drop(columns= [col]).reset_index(drop= True), tfidf_df.reset_index(drop= True)], axis= 1)
```

```python
X= df.drop(columns= ['Crm Cd'])
y= df['Crm Cd']

target_labelencoder= LabelEncoder()
y_encoded= target_labelencoder.fit_transform(y)

cat_cols= X.select_dtypes(include= ['object', 'category']).columns
X_encoded= X.copy()
X_test_encoded= X_test.copy()
encoders= {}
for col in cat_cols:
    le= LabelEncoder()
    X_encoded[col]= le.fit_transform(X[col].astype(str))
    X_test_encoded[col]= le.transform(X_test[col].astype(str))
    encoders[col]= le

X_train, X_val, y_train, y_val= train_test_split(X_encoded, y_encoded, random_state= 48, test_size= 0.2, stratify= y_encoded)
```

```python
cols_to_select= X_encoded.columns.to_list()[15:]
cols_to_keep= X_encoded.columns.to_list()[:15]
X_select_part= X_encoded.iloc[:, 15:]
X_keep_part= X_encoded.iloc[:, :15]
selector= SelectKBest(score_func= chi2, k= 200)
X_select_transformed= selector.fit_transform(X_select_part, y_encoded)
selected_indices= selector.get_support(indices= True)
selected_features= [cols_to_select[i] for i in selected_indices]
X_selected_df= pd.DataFrame(X_select_transformed, columns= selected_features, index= X_encoded.index)
X_encoded= pd.concat([X_keep_part, X_selected_df], axis= 1)
def transform_other(X_other):
    X_other_select= X_other.iloc[:, 15:]
    X_other_keep= X_other.iloc[:, :15]
    X_other_selected= selector.transform(X_other_select)
    X_other_selected_df= pd.DataFrame(X_other_selected, columns= selected_features, index= X_other.index)
    return pd.concat([X_other_keep, X_other_selected_df], axis=1)
X_train= transform_other(X_train)
X_val= transform_other(X_val)
X_test= transform_other(X_test_encoded)
```

```python
lr_model= LogisticRegression(max_iter= 1000, random_state= 48)
# param_grid= {
#     'max_iter': [100, 1000]
# }
# grid= GridSearchCV(lr_model, param_grid, cv= 4, scoring= 'f1_macro', n_jobs= -1, verbose= 2)
# grid.fit(X_train, y_train)
# print(grid.best_params_)
# predict= grid.best_estimator_.predict(X_val)
lr_model.fit(X_train, y_train)
predict= lr_model.predict(X_val)
f1_score(y_val, predict, average= 'macro')
```

```python
rf_model= RandomForestClassifier(n_estimators= 1000, max_depth= 100, random_state= 48)
# param_grid= {
#     'n_estimators': [10, 100, 1000],
#     'max_depth': [5, 10, 100, 1000]
# }
# grid= GridSearchCV(rf_model, param_grid, cv= 4, n_jobs= -1, scoring= 'f1_macro', verbose= 2)
# grid.fit(X_train, y_train)
# print(grid.best_params_)
# predict= grid.best_estimator_.predict(X_val)
rf_model.fit(X_train, y_train)
predict= rf_model.predict(X_val)
f1_score(y_val, predict, average= "macro")
```
