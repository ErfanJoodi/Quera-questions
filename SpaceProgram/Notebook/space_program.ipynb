<div dir="rtl">
<p dir=rtl style="direction: rtl;text-align: justify;line-height:200%;font-family:vazir;font-size:medium">
    <font face="vazir" size=3>

یک تکنولوژی پیشران جدید فضایی مبتنی بر ریاضیات پیشرفته توسعه یافته‌است. به همین منظور می‌خواهیم ۱۵ منطقه آموزشی که در دو سال اخیر تحصیلی و در آخرین پایه تحصیلی بیشترین رشد معناداری را داشته‌اند انتخاب کنیم و به آن‌ها بودجه تحقیق و توسعه اختصاص دهیم. این مراکز آینده سفرهای فضایی را رقم خواهند زد.

با استفاده از یک آزمون آماری مناسب، ۱۵ منطقه برتر آموزشی در دو سال اخیر تحصیلی را که بیشترین پیشرفت معنادار را داشته‌اند و در متغیر `significant` ذخیره کنید.

</font>
    <br>
    <font face="vazir" color="red">
        خروجی:
    </font>
خروجی شما باید در قالب یک لیست پایتون به اندازه ۱۵ باشد که به ترتیب نزولی مرتب شده باشند. برای داشتن ترتیب نزولی، کافی‌ست لیست به‌صورتی باشد که در اولین عنصر لیست بهترین، عنصر دوم لیست برابر با دومین منطقه آموزشی از نظر پیشرفت معناداری و سومین عنصر برابر با سومین منطقه و ... باشد.
</p>
</div>

```python
df= data.copy()

# Ensure numeric columns
df['Proficient']= pd.to_numeric(df['Proficient'], errors= 'coerce').fillna(0).astype(int)
df['Total']= pd.to_numeric(df['Total'], errors= 'coerce').fillna(0).astype(int)

# Normalize School Year to sortable type if needed
# If School Year is string like "2020-2021", this still sorts lexicographically OK for relative order.
# If it's numeric, that's fine too.
years= df['School Year'].unique()
# ---------- PART 1: stable districts by Grade for each Topic ----------

def compute_stable_districts(df, topic):
    dd= df[df['Topic'] == topic].copy()
    # find grades that appear in at least 2 distinct years
    grade_year_counts= dd.groupby('Grade')['School Year'].nunique()
    allowed_grades= grade_year_counts[grade_year_counts >= 2].index.tolist()
    results= []
    for g in sorted(allowed_grades):
        sub= dd[dd['Grade'] == g].copy()
        # per District, compute yearly proficiency proportions
        # require district to have >=2 years of data in this grade
        grp= sub.groupby(['District', 'School Year']).agg({'Proficient':'sum','Total':'sum'}).reset_index()
        # compute proportion per year
        grp['prop']= grp['Proficient'] / grp['Total'].replace({0: np.nan})
        # compute variance and mean for each district over years
        stats= grp.groupby('District').agg(
            years_count= ('School Year','nunique'),
            var_prop= ('prop', 'var'),
            mean_prop= ('prop','mean')
        ).reset_index()
        # keep districts with >=2 years_count
        stats= stats[stats['years_count'] >= 2].copy()
        # if var_prop NaN (e.g., identical values or single), replace with 0
        stats['var_prop'] = stats['var_prop'].fillna(0)
        if stats.empty:
            # no district with >=2 years for this grade: skip (shouldn't happen if allowed_grades criteria used)
            continue
        # choose district with minimal variance; tie-breaker: higher mean_prop, then smaller District id
        stats= stats.sort_values(by= ['var_prop','mean_prop','District'], ascending= [True,False,True])
        chosen= int(stats.iloc[0]['District'])
        results.append({'Grade': g, 'District': chosen})
    out= pd.DataFrame(results, columns= ['Grade','District'])
    out= out.sort_values('Grade').reset_index(drop= True)
    # ensure columns types
    out['Grade']= out['Grade'].astype(df['Grade'].dtype)
    out['District']= out['District'].astype(int)
    return out

stable_math_districts= compute_stable_districts(df, topic= 'Math')
stable_reading_districts= compute_stable_districts(df, topic= 'Reading')

# Final naming consistent with problem statement:
stable_math_districts= stable_math_districts[['District','Grade']].rename(columns= {'District':'District','Grade':'Grade'})
stable_reading_districts= stable_reading_districts[['District','Grade']].rename(columns= {'District':'District','Grade':'Grade'})

## Part 2
# 1. Find the latest school year in the dataset
latest_year= df['School Year'].max()

# 2. Filter the DataFrame for the latest year
df_latest_year= df[df['School Year'] == latest_year].copy()

# 3. Calculate the total number of proficient students nationwide in the latest year
total_proficient_nationwide= df_latest_year['Proficient'].sum()

# 4. Group by district and sum their proficient students
district_proficient_sum= df_latest_year.groupby('District')['Proficient'].sum()

# 5. Calculate the probability for each district
probabilities= district_proficient_sum / total_proficient_nationwide

# 6. Create the final DataFrame
most_probable= probabilities.reset_index(name= 'Probability')

# 7. Sort by probability, select top 10, round the values, and reset the index
most_probable= most_probable.sort_values('Probability', ascending= False).head(10)
most_probable['Probability']= most_probable['Probability'].round(3)
most_probable= most_probable.reset_index(drop=True)

## Part 3
# 1. Identify the last two years and the latest grade
all_years= sorted(df['School Year'].unique())
latest_year= all_years[-1]
penultimate_year= all_years[-2]
latest_grade= df['Grade'].max()

# 2. Filter data for the specific conditions
growth_df= df[
    (df['Topic'] == 'Math') &
    (df['Grade'] == latest_grade) &
    (df['School Year'].isin([penultimate_year, latest_year]))
]

# 3. Pivot the table to align the two years' data for each district
pivot_df= growth_df.pivot_table(
    index= 'District',
    columns= 'School Year',
    values= ['Proficient', 'Total']
)

# 4. Drop districts that do not have data for both years
pivot_df.dropna(inplace= True)

# 5. Define a function to calculate the Z-score for growth between two proportions
def calculate_z_score(row):
    x1= row[('Proficient', penultimate_year)]
    n1= row[('Total', penultimate_year)]
    x2= row[('Proficient', latest_year)]
    n2= row[('Total', latest_year)]

    if n1 == 0 or n2 == 0: return -np.inf
    p1= x1 / n1
    p2= x2 / n2
    if p2 <= p1: return -np.inf # We only care about growth

    p_pooled= (x1 + x2) / (n1 + n2)
    if p_pooled == 0 or p_pooled == 1: return 0

    se= np.sqrt(p_pooled * (1 - p_pooled) * (1/n1 + 1/n2))
    if se == 0: return np.inf

    return (p2 - p1) / se

# 6. Apply the function to calculate Z-scores for each district
pivot_df['z_score']= pivot_df.apply(calculate_z_score, axis= 1)

# 7. Sort by Z-score to find the top 15 districts with the most significant growth
top_15_districts= pivot_df.sort_values('z_score', ascending= False).head(15)

# 8. Create the final list of district IDs
significants= top_15_districts.index.tolist()
```
